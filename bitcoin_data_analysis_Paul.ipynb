6...Does Volume Confirm Bitcoin Price Moves?

```python
data["Close"] - data["Open"]
```

```python
data["price_change"] = data["Close"] - data["Open"]
```

```python
data["Volume"]
```

```python
data["Volume_change"] = data["Volume"].pct_change()
```

```python
def classify_volume_price(row):
    if row["price_change"] > 0 and row["Volume_change"] > 0:
        return "Strong Bullish"
    elif row["price_change"] < 0 and row["Volume_change"] > 0:
        return "Strong Bearish"

    elif row["price_change"] > 0 and row["Volume_change"] < 0:
        return "Weak Bullish"
    else:
        return "Weak Bearish"
```

```python
data.apply(classify_volume_price, axis=1)
```

```python
data["volume_price_signal"] = data.apply(classify_volume_price, axis=1)
```

```python
data["Date"] = pd.to_datetime(data["Date"])
data["Year"] = data["Date"].dt.year
data["daily_return"] = data["Close"].pct_change()

yearly_returns = data.groupby("Year")["daily_return"].apply(
    lambda x: (1 + x.dropna()).prod() - 1
)

yearly_df = yearly_returns.reset_index()
yearly_df.columns = ["Year", "Return (%)"]

yearly_df["Return (%)"] = yearly_df["Return (%)"] * 100

print(yearly_df)
```

```python
print(yearly_df)
```

```python
print(yearly_df.to_string(index=False))
```

```python
yearly_df = yearly_returns.reset_index()
yearly_df.columns = ["Year", "Return (%)"]
yearly_df["Year"] = yearly_df["Year"].astype(str)

print(yearly_df)
```

```python
print(yearly_df.to_string(index=False))
```

```python
yearly_df["Return_Percent"] = yearly_df["Return (%)"] * 100
```

```python
print(yearly_df[["Year", "Return (%)"]].to_string(index=False))
print(yearly_df["Return (%)"].dtype)
print(yearly_df["Return (%)"].min(), yearly_df["Return (%)"].max())
```

```python
import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(go.Bar(
    x=yearly_df["Year"].astype(str).tolist(),
    y=yearly_df["Return (%)"].tolist(),
    name="Yearly Return"
))

fig.update_layout(
    title="Bitcoin Yearly Returns - Best vs Worst Years",
    xaxis_title="Year",
    yaxis_title="Return (%)",
    width=1200,
    height=600
)

fig.add_hline(y=0, line_dash="dash")
fig.show()
```
